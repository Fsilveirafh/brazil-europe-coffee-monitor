```python
import pandas as pd

raw = pd.read_excel(
    "../data/raw/cepea_arabica_daily.xls",
    engine="xlrd",
    skiprows=3,
    engine_kwargs={"ignore_workbook_corruption": True},
)
raw.head()
```

```python
raw.dtypes
```

```python
df = raw.copy()
df.columns = ["date", "price_brl_bag", "price_usd_bag"]

df["date"] = pd.to_datetime(df["date"], format="%d/%m/%Y")

for col in ["price_brl_bag", "price_usd_bag"]:
    df[col] = (
        df[col]
        .str.replace(".", "", regex=False)
        .str.replace(",", ".", regex=False)
        .astype(float)
    )

df.info()
df.tail()
```

```python
df.to_csv("../data/processed/cepea_arabica_daily_clean.csv", index=False)
```
